## Reinforcement Learning Example

In this example from Reinforcement Learning, the task is to use tools from Machine Learning to predict how an agent should act. We will then use those predictions to drive the behavior of the agent. Ideally, our intelligent agent should get a much better score than a random agent.

## Key concepts:

- **Observation**: These are the states of the game. It describes where the agent currently is.
- **Action**: These are the moves that the agent makes.
- **Episode**: One full game played from beginning (`env.reset()`) to end (when `done == True`).
- **Step**: Part of a game that includes one action. The game transitions from one observation to the next.

# 第1部分：收集数据

我们想要构建一个智能的决策体（actor），但首先需要收集哪些动作是有用的数据。

参考上面的代码，使用一个**随机**智能体在环境中运行至少 1000 个回合（episode），并记录每一步的数据。

在生成完这些数据后，建议再添加一列（例如 `total_reward`），表示该回合的累计总奖励。

完成数据收集后，你可以使用 Pandas（或类似工具）计算随机智能体**每个回合**的平均总奖励。这个平均分数通常在 1–2% 左右，说明随机智能体几乎不会赢得游戏。

## 提示

* `initial_observation = env.reset()` 会重置环境并返回该回合的初始观测值。
* `new_observation, reward, done, info = env.step(new_action)` 会执行 `new_action` 并返回下一个观测值、新的奖励、是否结束标志 `done` 以及额外信息 `info`。
* 当 `done != True` 时，说明回合尚未结束。
* 我们的目标是**最大化每回合的累计奖励**。如果智能体未能到达目标，通常该回合的总奖励为 0。
* `env.action_space.n` 给出环境中可选动作的总数。`env.action_space.sample()` 可以随机采样一个动作。
* `env.observation_space.n` 给出环境中可观测状态的总数。


````python
env = gym.make('FrozenLake-v1')

num_episodes = 40000

life_memory = []
for i in range(num_episodes):
    
    # start a new episode and record all the memories
    old_observation = env.reset()
    done = False
    tot_reward = 0
    ep_memory = []
    while not done:
        new_action = env.action_space.sample()
        observation, reward, done, info = env.step(new_action)
        tot_reward += reward
        
        ep_memory.append({
            "observation": old_observation,
            "action": new_action,
            "reward": reward,
            "episode": i,
        })
        old_observation = observation
        
    # incorporate total reward
    num_steps = len(ep_memory)
    for i, ep_mem in enumerate(ep_memory):
        ep_mem["tot_reward"] = tot_reward
        ep_mem["decay_reward"] = i*tot_reward/num_steps
        
    life_memory.extend(ep_memory)
    
memory_df = pandas.DataFrame(life_memory)
````

由此，你需要为**每一个观测—动作对**设计一组特征，并让你的模型去预测一个量。这个“量”应该是什么？可以是：

* **即时奖励**（immediate reward）：模型预测每个动作一步之后能得到的奖励。但注意：在回合开始时，大多数动作的即时奖励可能都是 0，这种情况下模型如何区分哪些动作更有利于最终到达目标？
* **累计回合总奖励**（total episode reward）：预测从当前状态执行某个动作开始，到回合结束能获得的**累计奖励**。这需要你对数据做一些分组——每条记录不仅包含一步的观测与动作，还要关联到该回合后续的总奖励。
* **走到终点的概率**（success probability）：模型预测执行该动作后最终胜利（到达目标）的可能性。
* **到目标的距离变化**：某些环境可以测量到离目标的欧氏距离或步数距离，你可以让模型预测该动作后与目标的距离减少了多少。

下面对这段强化学习风格的“模拟—学习—评估”代码做一个逐行详解，帮助你理解它的思路和每一步在做什么。

```python
from sklearn.ensemble import RandomForestRegressor

# 1. 拟合一个回归模型
model = RandomForestRegressor()

# 2. 构造训练目标 y
#    这里 y 是对多种奖励信号的线性组合：
#      - memory_df.reward      : 每一步的即时奖励
#      - memory_df.tot_reward  : 该回合的总奖励（同一个回合里每一步都一样）
#      - memory_df.decay_reward: 某种衰减奖励（假设越往后越不值钱）
y = (
    1.0 * memory_df.reward
  + 1.0 * memory_df.tot_reward
  + 0.1 * memory_df.decay_reward
)

# 3. 构造训练特征 x：当前状态（observation）和动作（action）
x = memory_df[["observation", "action"]]

# 4. 训练随机森林回归器，使它学会给定 (state, action) 预测一个“价值”y
model.fit(x, y)
```

```python
num_episodes = 500     # 要模拟的总回合数
random_per = 0         # ε–greedy 中 ε 的初始值（这里设 0，就一直用模型决策）

life_memory = []       # 用来存放所有回合的“步”数据
for i in range(num_episodes):
    # 每个回合开始
    old_observation = env.reset()  # 重置环境，获得初始状态
    done = False
    tot_reward = 0                 # 本回合累计总奖励
    ep_memory = []                 # 暂存本回合的所有步数据

    # 一步步走完整个回合
    while not done:
        if np.random.rand() < random_per:
            # ε–greedy：以 probability ε 随机探索一个动作
            new_action = env.action_space.sample()
        else:
            # 否则用训练好的 model 预测每个动作的 y 值，选择最大的
            # 这里 pred_in 会是四行：[(state,0), (state,1), (state,2), (state,3)]
            pred_in = [[old_observation, a] for a in range(env.action_space.n)]
            # model.predict 返回 4 个预测值，我们 pick argmax
            new_action = np.argmax(model.predict(pred_in))

        # 执行动作
        observation, reward, done, info = env.step(new_action)
        tot_reward += reward  # 累加本回合的总奖励

        # 记录这一步的数据
        ep_memory.append({
            "observation": old_observation,
            "action": new_action,
            "reward": reward,
            "episode": i,
        })

        # 更新状态，进入下一步
        old_observation = observation

    # 回合结束后：把该回合的总奖励附加到每一步的记录里
    for step in ep_memory:
        step["tot_reward"] = tot_reward

    # 把本回合的步数据加到全局记录里
    life_memory.extend(ep_memory)

# 用 pandas 把所有回合的数据拼成一个 DataFrame
memory_df2 = pandas.DataFrame(life_memory)
```

## 整体思路

1. **离线学习**：先收集一个 `memory_df`，里面包含每一步的（状态、动作、即时奖励、回合总奖励、衰减奖励等），然后训练一个回归模型，让它学会“在某个状态下采取某个动作，最终能得到怎样的回合回报”。
2. **策略模拟**：用训练好的模型做策略（贪心地选预测回报最高的动作），再跑一次多回合，生成新的 `memory_df2`。
3. **评估**：统计 `memory_df2` 中每个回合的累计奖励，计算平均值，看看模型策略到底效果如何。

这种做法相当于**通过随机或其他策略先生成数据，再做监督学习拟合一个“价值函数”，然后再用这个价值函数来指导决策**。它并不是标准的强化学习算法，但在简单环境下能快速上手并做初步策略改进。
