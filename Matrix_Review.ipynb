   - `A[0,0]`, `A[1,0]` 分别是矩阵第一列的两个分量，也就是向量 \(a_1=(A_{0,0},\,A_{1,0})\)。  
   - `scale=30`：缩放因子，使得箭头长度更易于观看。  
   - `label="$a_{1}$"`：图例标签。

   ax.quiver([0, 0], [0, 0], A[0,1], A[1,1],

   plt.title("columns of $A$")

A common occurance is a matrix with more columns than rows, but in data, more rows than columns is more common. Consider the matrix $F$.


````python
F=np.array([[1,2],[1,-2],[-1,1]])
Matrix(F)
````

Although, the rows of matrix $F$ are 3 dimensions, you can only describe the point lying in the 2D plane, as shown here:


````python
ax = plt.figure().add_subplot(projection='3d')
p=null_space(F.T)
xx, yy = np.meshgrid(np.arange(-3,3,0.1), np.arange(-3,3,0.1))
z=(p[0]*xx+p[1]*yy)/p[2]
ax.plot_surface(xx, yy, z, alpha=0.1)
ax.quiver([0,0], [0,0], [0,0], F[0,:], F[1,:], F[2,:])

ax.set_xlim([-3, 3])
ax.set_ylim([-3, 3])
ax.set_zlim([-3, 3])
plt.show()
````

这段代码在 3D 空间里可视化了矩阵 \(F\) 的列向量以及与之正交的一个平面，具体步骤如下：

1. **创建 3D 坐标系**  
   ```python
   ax = plt.figure().add_subplot(projection='3d')
   ```  
   - 新建一个 Figure 并在其中创建一个带 `projection='3d'` 的子图 `ax`，用于绘制 3D 图形。

2. **计算与 \(F\) 列空间正交的法向量**  
   ```python
   p = null_space(F.T)
   ```  
   - `null_space(F.T)` 返回 \(F^T\) 的零空间基底向量数组 `p`，也就是满足 \(F^T p = 0\) 的向量——它正好与 \(F\) 的所有列都正交。

3. **构建平面网格**  
   ```python
   xx, yy = np.meshgrid(np.arange(-3,3,0.1),
                        np.arange(-3,3,0.1))
   z = (p[0]*xx + p[1]*yy) / p[2]
   ```  
   - 在 \([-3,3]\times[-3,3]\) 区域内，用步长 0.1 生成 \(x,y\) 网格。  
   - 平面方程来自 \(\;p_0 x + p_1 y + p_2 z = 0\)，解得 \(z = -(p_0 x + p_1 y)/p_2\)（代码里符号或可正可负，取决于 `p` 的方向）。

4. **绘制这个平面**  
   ```python
   ax.plot_surface(xx, yy, z, alpha=0.1)
   ```  
   - 用半透明（`alpha=0.1`）的网格面显示该平面，直观地看到它在 3D 空间中的位置。

5. **绘制 \(F\) 的列向量**  
   ```python
   ax.quiver([0,0], [0,0], [0,0],
             F[0,:], F[1,:], F[2,:])
   ```  
   - `quiver` 从原点 \((0,0,0)\) 出发，按 \((F_{0,j},F_{1,j},F_{2,j})\) 的分量画出每一列向量 \(F[:,j]\)。

6. **设置坐标范围并显示**  
   ```python
   ax.set_xlim([-3, 3])
   ax.set_ylim([-3, 3])
   ax.set_zlim([-3, 3])
   plt.show()
   ```  
   - 将三个轴都限在 \([-3,3]\) 范围内，确保平面和向量都清晰可见。  
   - `plt.show()` 渲染并展示图像。

---

**整体效果**：  
- 你会看到一张半透明的平面（ \(F^T\) 的零空间），以及若干从原点射出的箭头（ \(F\) 的列向量）。  
- 这清晰地演示了线性代数中“列空间（column space）” 与“左零空间（left null space）” 的几何关系：列向量都躺在平面的正交补空间之外，而平面正交于它们。

Therefore, the rank is 2.

### Frobenius Norm of a Matrix


The Frobenius norm, sometimes also called the **Euclidean norm**, is the matrix norm of an $m×n$. Matrix A is defined as the square root of the sum of the absolute squares of its elements:



Let's use the matrix A again as an example:
假设我们之前的矩阵  
\[
A = \begin{pmatrix}
a_{11} & a_{12} \\
a_{21} & a_{22}
\end{pmatrix}
\]  
那么它的 Frobenius（欧几里得）范数就是把所有元素平方后相加再开根号：

\[
\|A\|_F
= \sqrt{\,a_{11}^2 + a_{12}^2 + a_{21}^2 + a_{22}^2\,}.
\]

---

### 具体例子

比如，若  
\[
A = \begin{pmatrix}
1 & 2 \\
3 & 4
\end{pmatrix},
\]  
那么

\[
\|A\|_F
= \sqrt{1^2 + 2^2 + 3^2 + 4^2}
= \sqrt{1 + 4 + 9 + 16}
= \sqrt{30}
\approx 5.477.
\]

---

### 用 NumPy 一行搞定

```python
import numpy as np

A = np.array([[1, 2],
              [3, 4]])
fro_norm = np.linalg.norm(A, 'fro')
print(fro_norm)  # → 5.477225575051661
```

这样就能快速得到任意矩阵的 Frobenius 范数。

````python
Matrix(A), norm(A)
````

You could also calculate the matrix norm manually, using the formula provided above, and confirm that the results are the same.


````python
m, n = A.shape[0], A.shape[1] # get number of rows and columns 
ss = 0

for i in range(m):
    for j in range(n):
        ss += A[i,j] ** 2
np.sqrt(ss)
````

### Matrix Addition


Consider matrix $B$:


````python
B=np.array([[1,1],[1,-1]])
Matrix(B)
````

In general, if we have matrix $\mathbf{A}$ and matrix $\mathbf{B}$


$\begin{align}
\mathbf{A}= \begin{bmatrix}
a\_{11} & a\_{12} & \cdots & a\_{1n} \\\\\\
a\_{21} & a\_{22} & \cdots & a\_{2n} \\\\\\
\vdots & \vdots & \ddots & \vdots \\\\\\
a\_{m1} & a\_{m2} & \cdots & a\_{mn} \\\\\\
\end{bmatrix}
\end{align}$
$\begin{align}
\mathbf{B}=\begin{bmatrix}
b\_{11} & b\_{12} & \cdots & b\_{1n} \\\\\\
b\_{21} & b\_{22} & \cdots & b\_{2n} \\\\\\
\vdots & \vdots & \ddots & \vdots \\\\\\
b\_{m1} & b\_{m2} & \cdots & b\_{mn} \\\\\\
\end{bmatrix}
\end{align}$


To be added, the two matrices must have an equal number of rows and columns, where we add the corresponding row and column element:


$\begin{align}
\mathbf{A}+\mathbf{B}=\begin{bmatrix}
a\_{11} + b\_{11} & a\_{12} + b\_{12} & \cdots & a\_{1n} + b\_{1n} \\\\\\
a\_{21} + b\_{21} & a\_{22} + b\_{22} & \cdots & a\_{2n} + b\_{2n} \\\\\\
\vdots & \vdots & \ddots & \vdots \\\\\\
a\_{m1} + b\_{m1} & a\_{m2} + b\_{m2} & \cdots & a\_{mn} + b\_{mn} \\\\\\
\end{bmatrix}
\end{align}$


In numpy, we can add two arrays as follows:


````python
C=A+B
Matrix(C)
````

This matrix is symmetric. This means $B=B^{T}$, we can verify this in Python:


````python
B_T=B.T
Matrix(B_T)
````

### Matrix and Vector Multiplication


We can also multiply a matrix by a vector and get a new vector. Consider the vector $\mathbf{x}$:


$
\mathbf{x}=
\begin{bmatrix}
x\_1 \\\\\\
x\_2 \\\\\\
\vdots \\\\\\
x_n
\end{bmatrix}
$


The following  $\mathbf{Ax}=\mathbf b$  matrix multiplication  is defined as the dot product  $\mathbf{x}$ with each row of $\mathbf{A}$ :


$
\begin{matrix}a\_{11}x\_1+\cdots + a\_{1n}x_n=b\_1
\\\\\ a\_{21}x\_1+\cdots + a\_{2n}x_n =b\_2
\\\\\ \vdots
\\\\\ a\_{m1}x\_1+\cdots + a\_{mn}x_n =b_m\end{matrix}
$


Where ,


$
\mathbf{b}=
\begin{bmatrix}
b\_1 \\\\\\
b\_2 \\\\\\
\vdots \\\\\\
b_m
\end{bmatrix}
$


In numpy:


````python
x=np.array([1,1])
````

````python
A=np.array([[-1,1],[1,2]])
````

````python
b=A@x
Matrix(b)
````

We can see that $b$ is a combination of the rows of $A$:

$\mathbf{b}=x\_{1}\mathbf{a}*{1}+\mathbf{x}*{2}a\_{2}$ or $\mathbf{b}$ is a rotation of  $\mathbf{x}$


````python
fig, ax = plt.subplots(figsize = (12, 7))
ax.quiver([0, 0],[0, 0],A[0,0], A[1,0],scale=10,label="$a_{1}$")
ax.quiver([0, 0],[0, 0],A[0,1], A[1,1],scale=10,label="$a_{2}$")
ax.quiver([0,0],[0,0],b[0], b[1],scale=10,label="b",color='r')
ax.quiver([0,0],[0,0],x[0], x[1],scale=10,label="x",color='b')
ax.set_xlim([-10,10])
ax.set_ylim([-5,10])
fig.legend()
plt.show()
````

### Multiplying Matrices


If a matrix $\mathbf{C}$ is the product of matrix $\mathbf{A}$ and matrix $\mathbf{B}$, then the $i-th$ row $j-th$ column is obtained by multiplying term-by-term the entries of the $i-th$ row of $\mathbf{A}$ and the $j-th$ column of $\mathbf{B}$, and summing these $n$ products. In other words, you can also think of the  $i-th$ row $j-th$ column as the dot product of the $i-th$ row of $\mathbf{A}$  and the $j-th$ column of $\mathbf{B}$:


$\mathbf{C}=\mathbf{AB}$

$=\begin{pmatrix}
a\_{11}b\_{11} +\cdots + a\_{1n}b\_{n1} & a\_{11}b\_{12} +\cdots + a\_{1n}b\_{n2} & \cdots & a\_{11}b\_{1p} +\cdots + a\_{1n}b\_{np} \\\\\\

## Factorization for PCA


We can only use  eigen decomposition if the matrix is full rank, but even if the matrix is full rank, the eigenvalues and eigenvectors could be complex.

A useful application of the symmetric matrix ($S=S^{T}$) is where we want the eigenvalues to be real and the eigenvectors to be orthogonal. For example, the covariance matrix $C$ in PCA.
A few clarifications and then the “big picture” of why symmetry is so useful:

---

### 1. Eigen‐decomposition in general  
- **Any** \(n\times n\) square matrix \(A\) has up to \(n\) (possibly repeated or complex) eigenvalues and eigenvectors—you don’t need it to be full‑rank to *compute* them.  
- What *does* matter for writing  
  \[
    A = V\Lambda V^{-1}
  \]
  is that \(A\) be **diagonalizable**, i.e. have a full basis of \(n\) linearly independent eigenvectors.  
- If it isn’t diagonalizable (e.g. a defective matrix), you must fall back to a Jordan or Schur decomposition instead.

---

### 2. Why symmetric \(\Rightarrow\) real & orthogonal  
When \(S=S^T\) is real‐symmetric, the **Spectral Theorem** guarantees:

1. **All eigenvalues are real.**  
   \[
     S\,v = \lambda\,v
     \quad\Longrightarrow\quad
     v^T S v = \lambda\,v^Tv
     \quad\implies\quad
     \lambda\in\mathbb R
   \]
2. **Eigenvectors for distinct eigenvalues are orthogonal.**  
   If \(S\,u=\lambda_1u\) and \(S\,v=\lambda_2v\), then  
   \[
     u^T S v = \lambda_2 \,u^Tv,
     \quad
     (S u)^T v = \lambda_1\,u^T v
     \quad\Longrightarrow\quad
     (\lambda_1-\lambda_2)\,u^Tv=0.
   \]
3. **You can choose an orthonormal basis of eigenvectors**, i.e.  
   \[
     S = Q\,\Lambda\,Q^T,
     \quad Q^T Q = I,
     \quad \Lambda=\mathrm{diag}(\lambda_i).
   \]

---

### 3. Application to PCA  
- In PCA we form the **covariance matrix**  
  \[
    C = \frac1{N-1} X^T X,
  \]  
  which is by construction symmetric (and PSD).  
- By diagonalizing \(C = Q\Lambda Q^T\):  
  - The columns of \(Q\) are the principal directions (orthonormal!).  
  - The diagonal entries \(\lambda_i\ge0\) are the variances along those directions.  
- Because \(C\) is symmetric, you’re **guaranteed** to get real, non‑negative eigenvalues and a clean orthonormal basis—exactly what makes PCA numerically stable and interpretable.

---

**Bottom line**  
- You *can* compute eigenvalues/vectors of any square matrix, but only symmetric (or more generally normal/Hermitian) matrices enjoy the full‑strength spectral theorem: **real spectrum + orthogonal eigenvectors ↔ perfect diagonalization by a rotation**.  
- That’s why in data science we so often work with symmetric matrices (covariance, graph Laplacian, kernels, etc.)—they give you a guaranteed, well‑behaved eigendecomposition.

$\mathbf{C}=\mathbf{V}\mathbf{\Lambda}\mathbf{V}^{T}$


Using numpy,


````python
eigen_values , eigen_vectors = eig(C)
````

We can perform PCA using factorization. To find the first principle component, we need to find the eigenvector with the largest eigenvalue.


````python
v=eigen_vectors[:, np.argmax(eigen_values)].reshape(-1,1)
v
````

We can calculate first principal component by finding the projection of the original data onto $\mathbf{v}$.


$\mathbf{Z}=\mathbf{X} \mathbf{v}$


````python
Z=X@v
````
